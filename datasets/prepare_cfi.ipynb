This notebook is to inject data errors to the raw table chi_food_data.csv

```python
import pandas as pd
import random
from datetime import datetime
```

```python
file_path = 'chi_food_data.csv'
cfi_data = pd.read_csv(file_path)
cfi_data.head()
```

```python
# Error 1: Inconsistent formats in "Risk" column
risk_typos = {
    "Risk 1 (High)": ["Risk1(High)", "Risk 1 - High", "R1 High"],
    "Risk 2 (Medium)": ["Risk2(Medium)", "Risk 2 Medium", "R2 Medium"],
    "Risk 3 (Low)": ["Risk3(Low)", "Risk 3 Low", "R3 Low"]
}
cfi_data['Risk'] = cfi_data['Risk'].apply(
    lambda x: random.choice(risk_typos.get(x, [x])) if pd.notna(x) else x
)
```

```python
# Error 2: Typo and inconsistency in "Results" column
results_typos = {
    "Pass": ["PAss", "pass ", " Pass"],
    "Fail": ["fail", " FAIL", "Fail "],
    "Pass w/ Conditions": ["pass w/ cond", "PASS W/ CONDITIONS", "pass with conditions"]
}
cfi_data['Results'] = cfi_data['Results'].apply(
    lambda x: random.choice(results_typos.get(x, [x])) if pd.notna(x) else x
)
```

```python
# Error 3: Messy "City State Zip" column
cfi_data['City State Zip'] = cfi_data['City State Zip'].apply(
    lambda x: x.lower() if pd.notna(x) and random.random() < 0.5 else x
)
```

```python
# Error 4: Inconsistent "Inspection Date" formats
def random_date_format(date_str):
    try:
        date_obj = datetime.strptime(date_str, "%m/%d/%Y")
        formats = ["%Y-%m-%d", "%d-%m-%Y", "%m/%d/%Y"]
        return date_obj.strftime(random.choice(formats))
    except:
        return date_str

cfi_data['Inspection Date'] = cfi_data['Inspection Date'].apply(
    lambda x: random_date_format(x) if isinstance(x, str) else x
)

# Error 5: Typos in "Facility Type" column
facility_typos = {
    "Restaurant": ["Resturant", "Restaurent", "Restarunt"],
    "Grocery Store": ["Grocry Store", "GroceryStore", "Grocerie Store"]
}
cfi_data['Facility Type'] = cfi_data['Facility Type'].apply(
    lambda x: random.choice(facility_typos.get(x, [x])) if pd.notna(x) else x
)
```
