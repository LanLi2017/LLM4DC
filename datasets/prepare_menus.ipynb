This notebook is to inject data errors to the raw table menu.csv

```python
import pandas as pd
import random
from datetime import datetime
```

```python
file_path = 'menu_data.csv'
menu_data = pd.read_csv(file_path)
menu_data.head()
```

```python
# Error 1: uppercase format, introduce inconsistent cases in 'physical_description'
menu_data['physical_description'] = menu_data['physical_description'].apply(
        lambda x: x.lower() if isinstance(x, str) and random.random() < 0.5 else x
    )

menu_data['physical_description'].head()
```

```python
# Error 2: Transform dates (introduce non-ISO formats in 'date')
def random_date_format(date_str):
    try:
        date_obj = datetime.strptime(date_str, "%Y-%m-%d")
        formats = ["%m/%d/%Y", "%d-%m-%Y", "%Y.%m.%d"]
        return date_obj.strftime(random.choice(formats))
    except:
        return date_str

menu_data['date'] = menu_data['date'].apply(
    lambda x: random_date_format(x) if isinstance(x, str) and random.random()<0.4 else x)

menu_data['date'].head(10)
```

```python
# Error 3: Numeric transformation (introduce commas or extra characters in dish_count and page_count)
menu_data['dish_count'] = menu_data['dish_count'].apply(
    lambda x: f"{x } dishes" if random.random() < 0.7 else x
)
menu_data['page_count'] = menu_data['page_count'].apply(
    lambda x: f"{x} pages" if random.random() < 0.6 else x
)

menu_data['dish_count'].head(10)
```

```python
# Error 4: Regular-expression transformation (introduce typos or inconsistent formats in 'venue')
menu_data['venue'] = menu_data['venue'].apply(
    lambda x: f"{x}.?" if isinstance(x, str) and random.random() < 0.3 else x
)
menu_data['venue'].head()
```

```python
# Function to inject typos for the word "BANQUET"
def inject_typos(value):
    if isinstance(value, str) and "BANQUET" in value.upper():
        typos = ["banquuet", "banuqet", "banquet ", "banqu3t"]
        return random.choice(typos) if random.random() < 0.5 else value
    return value

# Apply the typo injection to the 'event' column
menu_data['event'] = menu_data['event'].apply(inject_typos)
menu_data['event'].head(50)
```

```python
# Error 5: Mass edit (introduce duplicates in 'sponsor')
menu_data['sponsor'] = menu_data['sponsor'].apply(
    lambda x: x.lower() if isinstance(x, str) and random.random() < 0.2 else x
)
menu_data['sponsor'] = menu_data['sponsor'].apply(
    lambda x: f"[{x}]" if isinstance(x, str) and random.random() < 0.5 else x
)
```

```python

# Error 6: Trim whitespaces (add extra spaces in 'occasion' and 'notes')
menu_data['occasion'] = menu_data['occasion'].apply(
    lambda x: f"  {x}  " if isinstance(x, str) and random.random() < 0.3 else x
)
menu_data['notes'] = menu_data['notes'].apply(
    lambda x: f" {x} " if isinstance(x, str) and random.random() < 0.3 else x
)
```

```python
# Function to inject errors and typos into the 'currency' column
def inject_currency_errors(value):
    if pd.isna(value):
        # Replace NaN with typo-like values
        typos = ["unknown", "n/a", "null", "None", "N/A"]
        return random.choice(typos) if random.random() < 0.7 else value
    elif isinstance(value, str):
        # Introduce typos in existing currency values
        typo_variants = {"Dollars": ["Dolars", "Doller", "Dollers"], 
                         "Pounds": ["Punds", "Pond", "Pounnd"]}
        for correct, typo_list in typo_variants.items():
            if correct.lower() in value.lower():
                return random.choice(typo_list) if random.random() < 0.5 else value
    return value

# Apply the typo injection to the 'currency' column
menu_data['currency'] = menu_data['currency'].apply(inject_currency_errors)
```
