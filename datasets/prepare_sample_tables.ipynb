```python
import pandas as pd
import random
from datetime import datetime
```

```python
# Error injection functions
def inject_trim_errors(value):
    if isinstance(value, str):
        return f"  {value}  " if random.random() < 0.3 else value
    return value

def inject_upper_errors(value):
    if isinstance(value, str):
        return value.lower() if random.random() < 0.3 else value
    return value

def inject_date_errors(value):
    try:
        date_obj = datetime.strptime(value, "%Y-%m-%d")
        formats = ["%m/%d/%Y", "%d-%m-%Y", "%Y/%m/%d"]
        return date_obj.strftime(random.choice(formats)) if random.random() < 0.3 else value
    except:
        return value

def inject_numeric_errors(value):
    if isinstance(value, (int, float)):
        return f"{value:,}" if random.random() < 0.3 else value
    return value

def inject_regex_errors(value):
    if isinstance(value, str):
        return value.replace(" ", "_") if random.random() < 0.3 else value
    return value

def inject_mass_edit_errors(value):
    if isinstance(value, str):
        return value.lower() if random.random() < 0.3 else value
    return value
```

```python
# Error injection mapping
error_injection_map = {
    "trim": inject_trim_errors,
    "upper": inject_upper_errors,
    "date": inject_date_errors,
    "numeric": inject_numeric_errors,
    "regexr_transform": inject_regex_errors,
    "mass_edit": inject_mass_edit_errors
}
```

```python
# Inject errors into relevant columns for each sample table
sample_tables = []
for i, columns in enumerate(relevant_columns):
    sample = new_cfi_data.sample(n=10, random_state=i)  # Exactly 10 rows for each table
    for column in columns:
        if column in sample.columns:
            operations = ["trim", "upper", "date", "numeric", "regexr_transform", "mass_edit"]
            for operation in operations:
                sample[column] = sample[column].apply(error_injection_map[operation])
    sample_tables.append(sample)

# Save each sample table to a CSV file
pp_ids = [35,43,44,45,46,47,48,50,51,53,54,55,56,57,58,59,60]
for i, table in enumerate(sample_tables):
    pp_id = pp_ids[i]
    output_path = f'cfi_updates/chi_food_data_p{pp_id}.csv'
    table.to_csv(output_path, index=False)
    print(f"Sample table {i+1} saved to {output_path}")
```

```python
# Enhanced error injection functions
def inject_trim_errors(value):
    if isinstance(value, str):
        # Add extra spaces randomly
        return f"  {value.strip()}  " if random.random() < 0.5 else value
    return value

def inject_upper_errors(value):
    if isinstance(value, str):
        # Randomly convert to lowercase or inconsistent casing
        return value.upper() if random.random() < 0.5 else value.lower()
    return value

def inject_date_errors(value):
    try:
        # Ensure the value can be parsed as a time in "HHMM" format
        date_obj = datetime.strptime(value, "%H%M")

        # Generate random incorrect formats or valid alternatives
        formats = [
            "%H:%M",         # Add colon to split hours and minutes
            "%I:%M %p",      # Use 12-hour format with AM/PM
            "%H%M",          # Original format
            "%H-%M",         # Replace colon with a dash
            "Hour: %H Min: %M",  # Add extra text to describe time
            "%M/%H",         # Swap minutes and hours
        ]

        # Randomly apply one of the valid/incorrect formats
        if random.random() < 0.7:
            return date_obj.strftime(random.choice(formats))
        else:
            # Inject truly messy values in 30% of cases
            messy_values = [
                "2400",          # Invalid time
                "99:99",         # Clearly incorrect format
                "Invalid Time",  # Non-numeric invalid time
                "",              # Missing value
                "12:00 midnight", # Full descriptive time
                None             # Null value
            ]
            return random.choice(messy_values)
    except:
        return value  # Return the original value if parsing fails

def inject_regex_errors(value):
    if isinstance(value, str):
        # Replace spaces with underscores or other special characters
        return value.replace(" ", "_") if random.random() < 0.5 else value.replace(" ", "-")
    return value

def inject_mass_edit_errors(value):
    if isinstance(value, str):
        # Normalize to lowercase or replace with similar strings
        replacements = {"aa": "AA", "Aa": "aa", "DL": "dl"}
        return replacements.get(value.lower(), value.lower()) if random.random() < 0.5 else value
    return value

def inject_missing_values(value):
    # Randomly introduce missing values
    return None if random.random() < 0.2 else value

def inject_src_tricky_errors(value):
    # Replace airline codes with full names or similar alternatives
    tricky_replacements = {
        "ord": ["O'Hare International", "ORD Intl.", "Chicago ORD"],
        "ua": ["United Airlines", "United Air", "UA Intl."],
        "flightrrrival": ["Flight Arr.", "Flight Arrival Service", "FArrival"],
        "usatoday": ["USA Today", "USAT", "U.S.A. Today"],
        "flylouisville": ["Louisville Airport", "FlyLouisville.com", "FLY Lou"],
        "helloflight": ["HLight", "Hello Light", "H.Of.Light"]
    }

    if isinstance(value, str) and value in tricky_replacements:
        # Replace with a random choice from the tricky options
        return random.choice(tricky_replacements[value])
    return value
```

```python
flights_ids = list(range(111,127))
# Sample and inject errors
# Sample and save clean and dirty tables
clean_sample_paths = []
dirty_sample_paths = []

for i, columns in enumerate(relevant_columns_for_errors):
    # Sample 50 rows for the current version
    clean_sample = flights_clean.sample(n=50, random_state=i)
    dirty_sample = clean_sample.copy()

    # Inject errors into the dirty sample
    for column in columns:
        if column in dirty_sample.columns:
            for func in fl_error_map.values():
                if column == "src" and func == "src_tricky":
                    dirty_sample[column] = dirty_sample[column].apply(func)
                else:
                    dirty_sample[column] = dirty_sample[column].apply(func)
    
    # Save the clean and dirty sample tables
    flight_id = flights_ids[i]
    dirty_path = f'flights/flights_data_p{flight_id}.csv'
    clean_path = f'flights/clean_tables/flights_sample_p{flight_id}.csv'
    clean_sample.to_csv(clean_path, index=False)
    dirty_sample.to_csv(dirty_path, index=False)

    clean_sample_paths.append(clean_path)
    dirty_sample_paths.append(dirty_path)

print("Clean sample tables saved as:", clean_sample_paths)
print("Dirty sample tables saved as:", dirty_sample_paths)
```
